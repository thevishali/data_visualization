```python
# ============================================================
# VEHICLE MARKET DATA VISUALIZATION
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. Load Dataset
# ------------------------------------------------------------

df = pd.read_csv("vehicle_market.csv")

print(df.head())


# ------------------------------------------------------------
# 2. Year vs Selling Price Scatter Plot
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="year",
    y="selling_price"
)

plt.title("Vehicle Year vs Selling Price")
plt.xlabel("Vehicle Year")
plt.ylabel("Selling Price")

plt.show()


# ------------------------------------------------------------
# 3. Selling Price Outlier Detection
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.boxplot(
    x=df["selling_price"]
)

plt.title("Selling Price Outlier Detection")

plt.show()


# ------------------------------------------------------------
# 4. Correlation Heatmap
# ------------------------------------------------------------

plt.figure(figsize=(10,7))

corr = df[
    ["year", "selling_price", "km_driven"]
].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Vehicle Market Correlation Heatmap")

plt.show()


# ------------------------------------------------------------
# 5. Fuel Type Categories
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="fuel"
)

plt.xticks(rotation=45)

plt.title("Vehicle Count by Fuel Type")

plt.xlabel("Fuel Type")
plt.ylabel("Number of Vehicles")

plt.show()


# ------------------------------------------------------------
# 6. Average Selling Price Over Vehicle Year
# ------------------------------------------------------------

year_price = df.groupby("year")["selling_price"].mean()

plt.figure(figsize=(12,5))

plt.plot(
    year_price.index,
    year_price.values
)

plt.title("Average Selling Price by Vehicle Year")

plt.xlabel("Vehicle Year")
plt.ylabel("Average Selling Price")

plt.show()


# ------------------------------------------------------------
# 7. Fuel Type Distribution
# ------------------------------------------------------------

fuel = df["fuel"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    fuel,
    labels=fuel.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Fuel Type Distribution")

plt.show()


# ------------------------------------------------------------
# 8. Top 10 Vehicle Brands/Names by Average Selling Price
# ------------------------------------------------------------

name_price = df.groupby("name")["selling_price"].mean()

name_price = name_price.sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(10,5))

name_price.plot(
    kind="bar"
)

plt.title("Top 10 Vehicles by Average Selling Price")

plt.ylabel("Average Selling Price")

plt.xlabel("Vehicle Name")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# 9. Selling Price Distribution
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.histplot(
    df["selling_price"],
    bins=30,
    kde=True
)

plt.title("Selling Price Distribution")

plt.xlabel("Selling Price")

plt.show()


# ------------------------------------------------------------
# 10. Pair Plot
# ------------------------------------------------------------

sns.pairplot(
    df[
        [
            "selling_price",
            "year",
            "km_driven"
        ]
    ]
)

plt.show()


# ------------------------------------------------------------
# 11. Bubble Chart: KM Driven vs Selling Price
# ------------------------------------------------------------

plt.figure(figsize=(9,6))

plt.scatter(
    df["km_driven"],
    df["selling_price"],
    s=df["year"] / 2,
    alpha=0.5,
    c=df["selling_price"],
    cmap="viridis"
)

plt.colorbar(
    label="Selling Price"
)

plt.xlabel("Kilometers Driven")

plt.ylabel("Selling Price")

plt.title("Bubble Chart: KM Driven vs Selling Price")

plt.show()


# ------------------------------------------------------------
# 12. Violin Plot - Selling Price by Fuel Type
# ------------------------------------------------------------

plt.figure(figsize=(10,6))

sns.violinplot(
    data=df,
    x="fuel",
    y="selling_price"
)

plt.xticks(rotation=45)

plt.title("Selling Price Distribution by Fuel Type")

plt.xlabel("Fuel Type")

plt.ylabel("Selling Price")

plt.show()


# ------------------------------------------------------------
# 13. Box Plot - Selling Price by Fuel Type
# ------------------------------------------------------------

plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="fuel",
    y="selling_price"
)

plt.xticks(rotation=45)

plt.title("Selling Price by Fuel Type")

plt.xlabel("Fuel Type")

plt.ylabel("Selling Price")

plt.show()


# ------------------------------------------------------------
# 14. Selling Price by Transmission
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="transmission",
    y="selling_price"
)

plt.title("Selling Price by Transmission")

plt.xlabel("Transmission")

plt.ylabel("Selling Price")

plt.show()


# ------------------------------------------------------------
# 15. Selling Price by Seller Type
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="seller_type",
    y="selling_price"
)

plt.xticks(rotation=45)

plt.title("Selling Price by Seller Type")

plt.xlabel("Seller Type")

plt.ylabel("Selling Price")

plt.show()


# ------------------------------------------------------------
# 16. Selling Price by Owner Type
# ------------------------------------------------------------

plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="owner",
    y="selling_price"
)

plt.xticks(rotation=45)

plt.title("Selling Price by Owner Type")

plt.xlabel("Owner Type")

plt.ylabel("Selling Price")

plt.show()


# ------------------------------------------------------------
# 17. KM Driven vs Selling Price
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="km_driven",
    y="selling_price"
)

plt.title("Kilometers Driven vs Selling Price")

plt.xlabel("Kilometers Driven")

plt.ylabel("Selling Price")

plt.show()


# ------------------------------------------------------------
# 18. Vehicle Count by Transmission
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="transmission"
)

plt.title("Vehicle Count by Transmission")

plt.xlabel("Transmission")

plt.ylabel("Number of Vehicles")

plt.show()


# ------------------------------------------------------------
# 19. Vehicle Count by Seller Type
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="seller_type"
)

plt.xticks(rotation=45)

plt.title("Vehicle Count by Seller Type")

plt.xlabel("Seller Type")

plt.ylabel("Number of Vehicles")

plt.show()


# ------------------------------------------------------------
# 20. Vehicle Count by Owner Type
# ------------------------------------------------------------

plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="owner"
)

plt.xticks(rotation=45)

plt.title("Vehicle Count by Owner Type")

plt.xlabel("Owner Type")

plt.ylabel("Number of Vehicles")

plt.show()


# ------------------------------------------------------------
# COMPLETED
# ------------------------------------------------------------

print("Vehicle Market Data Visualization Completed Successfully.")
```
