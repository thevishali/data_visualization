```python
# ============================================================
# VEHICLE MARKET DATA ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# ------------------------------------------------------------
# 1. Load Dataset
# ------------------------------------------------------------

df = pd.read_csv("vehicle_market.csv")

print(df.head())

print(df.shape)

print(df.dtypes)


# ------------------------------------------------------------
# 2. Statistical Summary
# ------------------------------------------------------------

print(df.describe())


# ------------------------------------------------------------
# 3. Missing Values
# ------------------------------------------------------------

print("Missing Values\n")

print(df.isnull().sum())


# ------------------------------------------------------------
# 4. High Selling Price Vehicles
# ------------------------------------------------------------

high_price = df[df["selling_price"] > df["selling_price"].quantile(0.75)]

print(high_price.head())


# ------------------------------------------------------------
# 5. Petrol Vehicles
# ------------------------------------------------------------

petrol = df[df["fuel"] == "Petrol"]

print(petrol.head())


# ------------------------------------------------------------
# 6. Sort Vehicles by Selling Price
# ------------------------------------------------------------

sorted_df = df.sort_values(
    by="selling_price",
    ascending=False
)

print(sorted_df.head())


# ------------------------------------------------------------
# 7. Mean, Median, Mode and Standard Deviation
# ------------------------------------------------------------

print("Mean :", df["selling_price"].mean())

print("Median :", df["selling_price"].median())

print("Mode :")

print(df["selling_price"].mode())

print("Standard Deviation :", df["selling_price"].std())


# ------------------------------------------------------------
# 8. Average Selling Price by Fuel Type
# ------------------------------------------------------------

fuel_avg = df.groupby("fuel")["selling_price"].mean()

print(fuel_avg.sort_values(ascending=False))


# ------------------------------------------------------------
# 9. Top Fuel Types by Average Selling Price
# ------------------------------------------------------------

fuel_avg.sort_values(ascending=False).plot(
    kind="bar"
)

plt.title("Average Selling Price by Fuel Type")

plt.ylabel("Average Selling Price")

plt.xlabel("Fuel Type")

plt.show()


# ------------------------------------------------------------
# 10. Average Selling Price by Transmission
# ------------------------------------------------------------

transmission_avg = df.groupby("transmission")["selling_price"].mean()

print(transmission_avg.sort_values(ascending=False))


# ------------------------------------------------------------
# 11. Transmission Price Comparison
# ------------------------------------------------------------

transmission_avg.sort_values(ascending=False).plot(
    kind="bar"
)

plt.title("Average Selling Price by Transmission")

plt.ylabel("Average Selling Price")

plt.xlabel("Transmission")

plt.show()


# ------------------------------------------------------------
# 12. Average Selling Price by Seller Type
# ------------------------------------------------------------

seller_avg = df.groupby("seller_type")["selling_price"].mean()

print(seller_avg.sort_values(ascending=False))


# ------------------------------------------------------------
# 13. Seller Type Price Comparison
# ------------------------------------------------------------

seller_avg.sort_values(ascending=False).plot(
    kind="bar"
)

plt.title("Average Selling Price by Seller Type")

plt.ylabel("Average Selling Price")

plt.xlabel("Seller Type")

plt.show()


# ------------------------------------------------------------
# 14. Average Selling Price by Owner Type
# ------------------------------------------------------------

owner_avg = df.groupby("owner")["selling_price"].mean()

print(owner_avg.sort_values(ascending=False))


# ------------------------------------------------------------
# 15. Owner Type Price Comparison
# ------------------------------------------------------------

owner_avg.sort_values(ascending=False).plot(
    kind="bar"
)

plt.title("Average Selling Price by Owner Type")

plt.ylabel("Average Selling Price")

plt.xlabel("Owner Type")

plt.show()


# ------------------------------------------------------------
# 16. Correlation Analysis
# ------------------------------------------------------------

corr = df[
    ["year", "selling_price", "km_driven"]
].corr()

print(corr)


# ------------------------------------------------------------
# 17. Correlation Heatmap
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Vehicle Market Correlation Heatmap")

plt.show()


# ------------------------------------------------------------
# 18. Year vs Selling Price
# ------------------------------------------------------------

sns.scatterplot(
    data=df,
    x="year",
    y="selling_price"
)

plt.title("Vehicle Year vs Selling Price")

plt.show()


# ------------------------------------------------------------
# 19. KM Driven vs Selling Price
# ------------------------------------------------------------

sns.scatterplot(
    data=df,
    x="km_driven",
    y="selling_price"
)

plt.title("Kilometers Driven vs Selling Price")

plt.show()


# ------------------------------------------------------------
# 20. Fuel Type Distribution
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="fuel"
)

plt.title("Vehicle Count by Fuel Type")

plt.xlabel("Fuel Type")

plt.ylabel("Number of Vehicles")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# 21. Transmission Distribution
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="transmission"
)

plt.title("Vehicle Count by Transmission")

plt.xlabel("Transmission")

plt.ylabel("Number of Vehicles")

plt.show()


# ------------------------------------------------------------
# 22. Seller Type Distribution
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="seller_type"
)

plt.title("Vehicle Count by Seller Type")

plt.xlabel("Seller Type")

plt.ylabel("Number of Vehicles")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# 23. Vehicle Price Distribution
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    df["selling_price"],
    kde=True
)

plt.title("Selling Price Distribution")

plt.xlabel("Selling Price")

plt.ylabel("Number of Vehicles")

plt.show()


# ------------------------------------------------------------
# 24. Selling Price Outliers
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["selling_price"]
)

plt.title("Selling Price Outliers")

plt.show()


# ------------------------------------------------------------
# 25. Km Driven Distribution
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    df["km_driven"],
    kde=True
)

plt.title("Kilometers Driven Distribution")

plt.xlabel("Kilometers Driven")

plt.ylabel("Number of Vehicles")

plt.show()


# ------------------------------------------------------------
# 26. Top 10 Most Expensive Vehicles
# ------------------------------------------------------------

top_10 = df.sort_values(
    by="selling_price",
    ascending=False
).head(10)

print(top_10[
    ["name", "year", "selling_price", "km_driven", "fuel"]
])


# ------------------------------------------------------------
# 27. Top 10 Most Driven Vehicles
# ------------------------------------------------------------

top_driven = df.sort_values(
    by="km_driven",
    ascending=False
).head(10)

print(top_driven[
    ["name", "year", "selling_price", "km_driven", "fuel"]
])


# ------------------------------------------------------------
# 28. Average Price by Vehicle Year
# ------------------------------------------------------------

year_avg = df.groupby("year")["selling_price"].mean()

print(year_avg.sort_values(ascending=False).head(10))


# ------------------------------------------------------------
# 29. Vehicle Year vs Average Selling Price
# ------------------------------------------------------------

year_avg.sort_index().plot(
    kind="line",
    figsize=(10, 5)
)

plt.title("Vehicle Year vs Average Selling Price")

plt.xlabel("Vehicle Year")

plt.ylabel("Average Selling Price")

plt.show()


# ------------------------------------------------------------
# 30. Final Dataset Information
# ------------------------------------------------------------

print("Final Dataset Shape :", df.shape)

print("\nFinal Missing Values:")

print(df.isnull().sum())

print("\nVehicle Market Data Analysis Completed.")
```
